# Первый датасет

```python
import pandas as pd
df_1 = pd.read_excel('/Users/david/Desktop/business-location-ai/business-location-ai/data/raw/stategov/2 Ввод в эксплуатацию жилых зданий (районы)_09.xls',
    sheet_name='период',
    header=None
)

# display(df_1.head(15))
df_1.head(25)
```

```python
df_1_clean = df_1.iloc[4:12].reset_index(drop=True)

df_1_clean['district'] = df_1_clean.iloc[:, 0]
df_1_clean['housing_completed_m2'] = df_1_clean.iloc[:, 17]

df_1_clean = df_1_clean[['district', 'housing_completed_m2']].copy()

df_1_clean
```

# Очистка второго датасета

```python
import pandas as pd
df_2 = pd.read_excel('/Users/david/Desktop/business-location-ai/business-location-ai/data/raw/stategov/4 Инвестиции в жилищное строительство по районам.xls',
    sheet_name='период',
    header=None
)

# display(df_1.head(15))
df_2.head(25)
```

```python
df_2_clean = df_2.iloc[5:13].reset_index(drop=True)

df_2_clean['district'] = df_2_clean.iloc[:,0]
df_2_clean['housing_investment_mln_kzt'] = df_2_clean.iloc[:,17]

df_2_clean = df_2_clean[['district', 'housing_investment_mln_kzt']].copy()
df_2_clean
```

```python
print(df_2_clean.shape)
print(df_2_clean.isna().sum())
print(df_2_clean["housing_investment_mln_kzt"].sum())
```

# Соединяю первый и второй датасет

```python
df_merged = pd.merge(
    df_1_clean,
    df_2_clean,
    on='district',# по какому столбцу соединять
    how="inner", # остаются только пересечения
    validate="one_to_one", # проверяет то что каждый район встречается только один раз в датасете
    indicator=True # добавляет колонку merge в которой можно посмотреть откуда пришли строки
)

print(df_merged.shape)
print(df_merged["_merge"].value_counts())
display(df_merged)
```

удалились некторые районы, надо проверить почему

```python
print("Только в первом:")
print(set(df_1_clean["district"]) - set(df_2_clean["district"]))

print("\nТолько во втором:")
print(set(df_2_clean["district"]) - set(df_1_clean["district"]))
```

там были пробелы из за этого изначально не получилось соединить

```python
df_1_clean["district"] = df_1_clean["district"].str.strip()
df_2_clean["district"] = df_2_clean["district"].str.strip()
```

```python
df_merged = pd.merge(
    df_1_clean,
    df_2_clean,
    on='district',# по какому столбцу соединять
    how="inner", # остаются только пересечения
    validate="one_to_one", # проверяет то что каждый район встречается только один раз в датасете
    indicator=True # добавляет колонку merge в которой можно посмотреть откуда пришли строки
)

print(df_merged.shape)
print(df_merged["_merge"].value_counts())
display(df_merged)
```
