# Feature Engineering sme_per_1000_residents

Если смотреть только на количество предприятий, район A лидирует. Но если нормализовать показатель на 1000 жителей:
- Район A: 100 МСП на 1000 жителей.
- Район B: 200 МСП на 1000 жителей.
Получается, относительно населения деловая активность в районе B выше.

Показывает сколько предприятиий приходится на 1000 людей

```python
dfv1 = df_merged_v3.copy()

dfv1['population_2026'] = dfv1['population_2026'].astype(int)
dfv1.info()
```

```python
dfv1['sme_per_1000_residents'] = (dfv1['active_sme_2026'] / dfv1['population_2026']) * 1000

dfv1[['district', 'active_sme_2026', 'population_2026', 'sme_per_1000_residents']].sort_values(by='sme_per_1000_residents', ascending=False)
```

# Housing Feature Engineering

Показывает сколько домов на 1000 человек

Высокое значение может указывать на активное жилищное строительство относительно численности населения. Это потенциально интересно для будущего спроса на кофейни.

```python
dfv1['housing_completed_m2'] = dfv1['housing_completed_m2'].astype(int)
dfv1.info()

dfv1['housing_m2_per_1000_residents'] = (dfv1['housing_completed_m2'] / dfv1['population_2026']) * 1000

dfv1[['district', 'housing_completed_m2', 'population_2026', 'housing_m2_per_1000_residents']].sort_values(by='housing_m2_per_1000_residents', ascending=False)
```

# Housing Investment Feature Engineering

Инвестиции в жилье на одного жителя


```python
dfv1['housing_investment_mln_kzt'] = dfv1['housing_investment_mln_kzt'].astype(int)
dfv1['housing_investment_mln_kzt'] = dfv1['housing_investment_mln_kzt']

# dfv1['housing_investment_mln_kzt']

# dfv1.info()

dfv1['housing_investment_per_capita_kzt'] = (dfv1['housing_investment_mln_kzt'] * 1000000 / dfv1['population_2026'])

dfv1[['district', 'housing_investment_mln_kzt', 'population_2026', 'housing_investment_per_capita_kzt']].sort_values(by='housing_investment_per_capita_kzt', ascending=False)
```

# EDA for 3 new features

Алатауский район лидирует по вводу жилья относительно населения, а Наурызбайский — по инвестициям на жителя.

Это может означать, что районы находятся на разных этапах строительной активности. Но подтвердить такую гипотезу по одному периоду статистики нельзя.

Особенно интересно, что Ауэзовский район имеет очень низкие показатели как ввода жилья, так и инвестиций. Это не обязательно плохой район для кофейни — возможно, он просто более застроенный и зрелый.

```python
import seaborn as sns
import matplotlib.pyplot as plt

order = dfv1.sort_values(by='sme_per_1000_residents',ascending=False)['district']

sns.barplot(data=dfv1, x='district', y ='sme_per_1000_residents', label='Скок предприятий на 1000 человек в одном районе', order=order)
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()

```

```python
order = dfv1.sort_values(by='housing_m2_per_1000_residents',ascending=False)['district']
sns.barplot(dfv1, x='district', y='housing_m2_per_1000_residents',label='Сколько домов на 1000 человек в одном районе', order=order)
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
order = dfv1.sort_values(by='housing_investment_per_capita_kzt',ascending=False)['district']
sns.barplot(dfv1, x='district', y='housing_investment_per_capita_kzt', label='Инвестиции в жилье на одного жителя', order=order)
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
